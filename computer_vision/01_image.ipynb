```python
# import libraries
import numpy as np
import cv2 as cv
```

```python
# read the image
img = cv.imread("./images/test_01.jpg")
```

```python
cv.imshow("images", img)
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
# check the size of image
img.shape
```

```python
# conver the image into grayscale and then check the shape

# convert the image into grayscale
gray_image = cv.cvtColor(img, cv.COLOR_BGR2GRAY)

# check the shape of gray_image
gray_image.shape
```

```python
# read webcam using cv
cap = cv.VideoCapture(0)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    gray_image = cv.cvtColor(frame, cv.COLOR_BGR2GRAY)
    cv.imshow("Grayscale camera", gray_image)

    if cv.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv.destroyAllWindows()
# change the cam into grayscale
```

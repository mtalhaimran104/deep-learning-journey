### Time series Weather Forecasting

```python
# time series library import
import os
import datetime
import pandas as pd
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import plotly.express as px
from statsmodels.tsa.seasonal import seasonal_decompose
```

### import the weather dataset

```python
# Download and extract the dataset
zip_path = tf.keras.utils.get_file(
    origin='https://storage.googleapis.com/tensorflow/tf-keras-datasets/jena_climate_2009_2016.csv.zip',
    fname='jena_climate_2009_2016.csv.zip',
    extract=True
)

# Get the extracted directory path
extracted_dir = os.path.splitext(zip_path)[0]

# Get the actual CSV file path
csv_path = os.path.join(
    extracted_dir,
    'jena_climate_2009_2016.csv'
)
```

```python
# Read the dataset into a Pandas DataFrame
df = pd.read_csv(csv_path)

# write the csv file
df.to_csv('jena_climate_2009_2016.csv')
```

```python
date_time = pd.to_datetime(df.pop('Date Time'), format='%d.%m.%Y %H:%M:%S')
```

```python
# plot the data
plot_cols = ['T (degC)', 'p (mbar)']
plot_features = df[plot_cols][:250]
plot_features.index = date_time[:250]
plot_features.plot(subplots=True)
```

```python
# decompose the data into trend, seasonality and noise
# we will use the seasonal_decompose function from statsmodels library to decompose the data into trend, seasonality and noise. We will use the 'T (degC)' column for this purpose.
decompose = seasonal_decompose(df['T (degC)'], model='additive', period=365)
decompose.plot()
```

```python
decompose = seasonal_decompose(df['rh (%)'], model='additive', period=365)
decompose.plot()
```

### Modeling

```python
# split the data into train, validation, and test. I have 35045 records make 70% for training and 20% for testing
# 70% of 35045 is 24531
# 20% of 35045 is 7009
# 10% of 35045 is 3505
# total 35045

# train data
train_df = df[:24531]
# validation data
val_df = df[24531:31540]
# test data
test_df = df[31540:]

# lets check the shape of the data
train_df.shape, val_df.shape, test_df.shape
```

```python
# lets plot the data
train_df.plot(subplots=True)
val_df.plot(subplots=True)
test_df.plot(subplots=True)
```

```python
# lets normalize the data
train_mean = train_df.mean()
train_std = train_df.std()

train_df = (train_df - train_mean) / train_std
val_df = (val_df - train_mean) / train_std
test_df = (test_df - train_mean) / train_std
```

```python
# lets plot the data
train_df.plot(subplots=True)
val_df.plot(subplots=True)
test_df.plot(subplots=True)
```
